```python
#create reference list for the median values of 'Review Rating' based on 'Category' - remove NA value from the calculation
category_median = df.groupby('Category')['Review Rating'].transform('median')
df['Review Rating'] = df['Review Rating'].fillna(category_median)
```

```python
df.isnull().sum()
```

```python
#replace header to snake case to better aligned with operating system
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
#replace to purchase amount usd to purchase_amount
df.columns = df.columns.str.replace('_(usd)','')
#or we can do df.columns = df.columns.str.lower().str.replace(' ','_')
df.columns
```

```python
# Create age_group column using qcut (4 equal-sized groups)
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)

# Create age_group_2 column using cut (Fixed 'bin=' to 'bins=')
df['age_group_2'] = pd.cut(df['age'], bins=[0, 25, 50, 100])

# Create age_group_3 column using cut (Fixed 'bin=' to 'bins=' and labels)
labels_3 = ['Young', 'Middle', 'Senior']
df['age_group_3'] = pd.cut(df['age'], bins=[0, 25, 50, 100], labels = labels_3)

# View all columns together side-by-side
df[['age', 'age_group', 'age_group_2', 'age_group_3']].head(10)
```

```python
#check the unique value and how many per each
df['frequency_of_purchases'].unique()
df['frequency_of_purchases'].value_counts()

# Define mapping dictionary (matching exact casing) 
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly':7,
    'Every 3 Months': 90,
    'Annually':365,
    'Quarterly' : 90,
    'Monthly': 30,
    'Bi-Weekly': 14
}
# Create new column purchase_frequency days
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
df[['discount_applied','promo_code_used']].head()
```

```python
#Check if discount_applied and promo_code_used is duplicated, if yes, we don't need both of them
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
#Remove promo_code_used as both of them are duplicated. By default, Pandas operates on rows (axis=0) so axis=1: Looks at Column Names.
df = df.drop('promo_code_used', axis=1)
df.columns
```
